```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np

# Optional: Set styles for better visuals
sns.set(style="whitegrid")
```

```python
#2. Basic Charts and Graphs
#1.1 Line Chart
        
x = np.arange(0, 10, 0.5)
y = np.sin(x)

plt.plot(x, y, marker='o')
plt.title('Line Chart of Sine Function')
plt.xlabel('X')
plt.ylabel('sin(X)')
plt.show()
```

```python
#2.2 Bar Chart

categories = ['A', 'B', 'C', 'D', 'E']
values = [23, 45, 56, 78, 33]

plt.bar(categories, values, color='skyblue')
plt.title('Bar Chart Example')
plt.xlabel('Categories')
plt.ylabel('Values')
plt.show()
      
```

```python
#2.3 Pie Chart
 
sizes = [15, 30, 45, 10]
labels = ['Frogs', 'Hogs', 'Dogs', 'Logs']

plt.pie(sizes, labels=labels, autopct='%1.1f%%', startangle=90)
plt.title('Pie Chart Example')
plt.show()
```

```python
#2.4 Histogram
 
data = np.random.randn(1000)

plt.hist(data, bins=30, color='purple', alpha=0.7)
plt.title('Histogram of Normally Distributed Data')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.show()
      
```

```python
#2.5 Scatter Plot

x = np.random.rand(100)
y = np.random.rand(100)

plt.scatter(x, y, color='orange')
plt.title('Scatter Plot Example')
plt.xlabel('X')
plt.ylabel('Y')
plt.show()
```

```python
#3. Intermediate Visualizations
#3.1 Multiple Lines in One Plot

x = np.linspace(0, 10, 100)
plt.plot(x, np.sin(x), label='sin(x)')
plt.plot(x, np.cos(x), label='cos(x)')
plt.legend()
plt.title('Multiple Line Plot')
plt.show()
```

```python
#3.2 Subplots

fig, axs = plt.subplots(2, 2, figsize=(10, 8))
axs[0, 0].bar(categories, values)
axs[0, 1].pie(sizes, labels=labels)
axs[1, 0].hist(data, bins=20)
axs[1, 1].scatter(x, y)

for ax in axs.flat:
    ax.label_outer()

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

dates = pd.date_range('20230101', periods=10)
series1 = np.random.randn(10).cumsum()
series2 = np.random.randn(10).cumsum()

plt.plot(dates, series1, label='Series 1')
plt.plot(dates, series2, label='Series 2')
plt.legend()
plt.title('Time Series Plot')
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Define x as an array of values
x = np.linspace(0, 2 * np.pi, 100)
# Define y as sine of x
y = np.sin(x)

# Plot the data
plt.plot(x, y)

# Add title and labels
plt.title('Customized Plot')
plt.xlabel('X-axis label')
plt.ylabel('Y-axis label')

# Add legend
plt.legend(['sin(x)'])

# Show grid
plt.grid(True)

# Save the plot to a file
plt.savefig('plot.png')
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

# Example dataset
data = pd.DataFrame({
    'category': ['A'] * 50 + ['B'] * 50,
    'value': np.concatenate([np.random.normal(0, 1, 50), np.random.normal(1, 1.5, 50)])
})

# Create violin plot
sns.violinplot(x='category', y='value', data=data)

# Show plot
plt.title("Violin Plot Example")
plt.show()
```
