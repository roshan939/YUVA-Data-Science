```python
import pandas as pd
import numpy as np

# Load dataset from CSV file

df = pd.read_csv('Online Purchase.csv')

# Display first few rows
print(df.head())
```

```python
# Basic info about dataset

print(df.info())
```

```python
# Statistical summary

print(df.describe(include='all'))
```

```python
# Check missing values
print(df.isnull().sum())

# Verify
#print(df_cleaned.isnull().sum())
```

```python
# Drop rows with missing values

df_cleaned = df.dropna()
print(df)
```

```python
# Handle missing values
# Fill missing numerical values with mean
df['Age'].fillna(df['Age'].mean(), inplace=True)
df['Salary'].fillna(df['Salary'].mean(), inplace=True)
print(df)
```

```python
# Rename columns for clarity
df.rename(columns={'Purchased': 'Purchase_Made'}, inplace=True)
print(df)
```

```python
# Group by Gender and get mean Age and Salary
grouped = df.groupby('Gender')[['Age', 'Salary']].mean()
print(grouped)
```

```python
# Sort data by Salary descending
sorted_df = df.sort_values(by='Salary', ascending=False)
print(sorted_df.head())
```

```python
# Correlation matrix
# Select only numeric columns
numeric_df = df.select_dtypes(include='number')

# Compute correlation matrix on numeric columns
corr = numeric_df.corr()
print(corr)
```

```python
 # Encode 'Gender' to numerical
df['Gender_Encoded'] = df['Gender'].map({'Male': 1, 'Female': 0})
print(df)
```

```python
# One-hot encoding for 'Gender'
df = pd.get_dummies(df, columns=['Gender'])
print(df)
```
